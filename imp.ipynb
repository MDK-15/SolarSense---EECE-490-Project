# Feature Importance Visualization

Loads a saved XGBoost model and plots per-feature importance scores as a bar chart. Useful for interpreting which input features drive each appliance's on/off predictions.

## Load Model

```python
import pickle
import pandas as pd
import matplotlib.pyplot as plt

# Load any model
# Change the model to wgat you need. For example: final_models/elec_television_on.pkl tests, the television model. final_models/elec_cooling_on.pkl test the cooling model.
with open('final_models/elec_television_on.pkl', 'rb') as f:
    model = pickle.load(f)
```

## Extract & Display Importances

```python
# Get feature importances as a sorted dataframe
importances = pd.Series(
    model.feature_importances_,
    index=model.feature_names_in_
).sort_values(ascending=False)

print(importances)
```

## Plot

```python
importances.plot(kind='bar', figsize=(10, 5))
plt.title('Feature Importances — elec_cooling_on')
plt.ylabel('Importance Score')
plt.tight_layout()
plt.show()
```
